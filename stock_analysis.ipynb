```python
import pandas as pd
```

```python
data = yf.download("AAPL",period="5y")
```

```python
data.head()
```

```python
tickers = [
    "AAPL", "MSFT", "GOOGL", "AMZN", "META", "TSLA", "JPM", "V", "JNJ", "NVDA",
    "RELIANCE.NS", "TCS.NS", "HDFCBANK.NS", "INFY.NS", "ICICIBANK.NS",
    "HINDUNILVR.NS", "SBIN.NS", "BAJFINANCE.NS", "WIPRO.NS", "AXISBANK.NS"
]

data = yf.download(tickers, period="10y")
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
data.shape
```

```python
data = data.dropna(how='all')
```

```python
data = data.ffill()
```

```python
data.isnull().sum()
```

```python
data.shape
```

```python
print(data.columns)
```

```python
data.columns = ['_'.join(col) for col in data.columns]
print(data.columns)
```

```python
print(data.index)
```

```python
data = data.reset_index()
```

```python
data.head()
```

```python
data_long = pd.melt(data, id_vars=['Date'], var_name='Ticker_Price', value_name='Value')
```

```python
data_long.head(10)
```

```python
data_long[['Price_Type', 'Ticker']] = data_long['Ticker_Price'].str.split('_', n=1, expand=True)
```

```python
data_long.head(10)
```

```python
data_long.head()
```

```python
data_long.shape
```

```python
data_long = pd.pivot(data_long, index=['Date','Ticker'], columns='Price_Type', values='Value')
```

```python
data_long.head()
```

```python
# 1. Reset the index
data_long = data_long.reset_index()

# 2. Safely flatten columns if it is a MultiIndex
if isinstance(data_long.columns, pd.MultiIndex):
    data_long.columns = [col[1] if col[1] != '' else col[0] for col in data_long.columns]
```

```python
data_long.head()
```

```python
data_long.shape
```

```python
data_long['Market'] = data_long['Ticker'].str.contains('.NS').map({True: 'India', False: 'US'})
```

```python
data_long.groupby('Market')['Ticker'].nunique()
```

```python
data_long['Daily_Return'] = data_long.groupby('Ticker')['Close'].pct_change() * 100
```

```python
data_long['Rolling_Avg_30'] = data_long.groupby('Ticker')['Close'].rolling(30).mean().reset_index(level=0, drop=True)
```

```python
data_long[data_long['Ticker']=='AAPL'].head(35)
```

```python
data_long['Volatility_30'] = data_long.groupby('Ticker')['Daily_Return'].rolling(30).std().reset_index(level=0, drop=True)
```

```python
print(data_long.shape)
print(data_long.columns.tolist())
print(data_long.isnull().sum())
```

```python
data_long.to_csv('stock_data_clean.csv', index=False)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
total_return = data_long.groupby('Ticker')['Close'].agg(['first', 'last'])
total_return['Return_Pct'] = (total_return['last'] - total_return['first']) / total_return['first'] * 100
```

```python
total_return.sort_values('Return_Pct', ascending=False)
```

```python
plt.figure(figsize=(14,6))
total_return_sorted = total_return.sort_values('Return_Pct', ascending=True)
plt.barh(total_return_sorted.index, total_return_sorted['Return_Pct'], color='steelblue')
plt.title('10 Year Total Return % by Stock')
plt.xlabel('Total Return %')
plt.axvline(x=0, color='red', linestyle='--')
plt.tight_layout()
plt.show()
```

```python
plt.savefig('total_return_by_stock.png', dpi=150, bbox_inches='tight')
```

```python
risk_reward = data_long.groupby('Ticker')[['Daily_Return', 'Volatility_30']].mean().reset_index()
```

```python
risk_reward.head()
```

```python
plt.figure(figsize=(10,7))
plt.scatter(risk_reward['Volatility_30'], risk_reward['Daily_Return'], color='steelblue', s=100)

for i, row in risk_reward.iterrows():
    plt.annotate(row['Ticker'], (row['Volatility_30'], row['Daily_Return']), 
                fontsize=8, ha='left')

plt.title('Risk vs Reward — 20 Global Stocks (10 Years)')
plt.xlabel('Average Volatility (Risk)')
plt.ylabel('Average Daily Return %')
plt.axhline(y=0, color='red', linestyle='--')
plt.savefig('risk_reward.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
returns_pivot = data_long.pivot(index='Date', columns='Ticker', values='Daily_Return')
```

```python
returns_pivot.shape
```

```python
corr_matrix = returns_pivot.corr()
```

```python
corr_matrix.shape
```

```python
plt.figure(figsize=(14,10))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Matrix — 20 Global Stocks Daily Returns')
plt.tight_layout()
plt.savefig('correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
top5 = total_return.sort_values('Return_Pct', ascending=False).head(5).index.tolist()
```

```python
top5
```

```python
top5_data = data_long[data_long['Ticker'].isin(top5)]
```

```python
top5_data.shape
```

```python
plt.figure(figsize=(14,7))

for ticker in top5:
    stock_data = top5_data[top5_data['Ticker'] == ticker]
    plt.plot(stock_data['Date'], stock_data['Close'], label=ticker)

plt.title('Top 5 Stocks — 10 Year Price Trend')
plt.xlabel('Date')
plt.ylabel('Close Price (USD/INR)')
plt.legend()
plt.tight_layout()
plt.savefig('top5_trend.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(12,6))

for ticker in top5:
    stock_data = top5_data[top5_data['Ticker'] == ticker]
    sns.kdeplot(stock_data['Daily_Return'].dropna(), label=ticker)

plt.title('Daily Return Distribution — Top 5 Stocks')
plt.xlabel('Daily Return %')
plt.ylabel('Density')
plt.legend()
plt.axvline(x=0, color='black', linestyle='--')
plt.tight_layout()
plt.savefig('return_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
summary = data_long.groupby(['Ticker', 'Market']).agg(
    Avg_Daily_Return=('Daily_Return', 'mean'),
    Avg_Volatility=('Volatility_30', 'mean'),
    Avg_Close=('Close', 'mean')
).reset_index()

summary = summary.merge(total_return[['Return_Pct']], on='Ticker')
summary = summary.round(2)
summary.to_csv('stock_summary.csv', index=False)
```
